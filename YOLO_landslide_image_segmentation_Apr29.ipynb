## Segmentation Prediction with trained weights

```python
#Prediction with trained weights
from ultralytics import YOLO
model = YOLO("/content/best.pt")  # load a trained model
```

```python
results = model("/content/Ching Cheung Road 2.png", conf=0.3) #Confidence level set to 0.3
```

```python
#Each row of box_np contains a predicted object. lower x coordinate of bounding box, lower y coordinate of bounding box, upper x coordinate of bounding box, confidence score, class label (1: landslide scar; 0: landslide debris)
box_np = results[0].boxes.cpu().data.numpy()
#Each element of mask_np contains a predicted mask.
mask_np = results[0].masks.cpu().data.numpy()
```

```python
import numpy as np
image_shape = results[0].boxes.cpu().orig_shape #image dimension
items = np.shape(box_np)[0]  #Number of predicted objects
```

```python
#Save each predicted objects into separate image
for i in range(items):
  new_boxes = torch.tensor(box_np[i])
  new_maskes = torch.tensor(mask_np[i])
  new_result = results[0].new()
  new_result.update(boxes=new_boxes, masks=new_maskes, probs=None, obb=None)
  new_result[0].save(filename="/content/result"+str(i+1)+".jpg")  # save to disk
```
